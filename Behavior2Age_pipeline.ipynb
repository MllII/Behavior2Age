**2.3 Поиск и анализ пропусков, дубликатов, выбросов**

```python
# ========== пропуски ==========
print("Пропуски по датафреймам:")
for name, df in dataframes.items():
    missing = df.isna().sum()
    missing = missing[missing > 0]
    if len(missing) == 0:
        print(f"{name}: пропусков нет")
    else:
        print(f"{name}:")
        display(missing.to_frame(name="missing_count"))
```

```python
# ========== полные дубликаты ==========
print("Полные дубликаты по датафреймам:")
for name, df in dataframes.items():
    dup_count = df.duplicated().sum()
    print(f"{name}: {dup_count}")
```

```python
# ========== проверка расхождения shape и nunique ==========
# Для users и ads_activity: shape > nunique(user_id), но полных дубликатов нет.
# Проверим, есть ли повторяющиеся user_id и конфликты значений.

for name, col in [("users", TARGET_COL), ("ads_activity", "ads_activity")]:
    df = dataframes[name]
    print("=" * 60)
    print(f"{name}")
    print("=" * 60)
    print(f"shape: {df.shape}")
    print(f"nunique({USER_ID_COL}): {df[USER_ID_COL].nunique()}")
    print(f"duplicated(subset=[user_id]).sum(): {df.duplicated(subset=[USER_ID_COL]).sum()}")

    dup_ids = df[df.duplicated(subset=[USER_ID_COL], keep=False)]
    print(f"Строк с повторяющимся user_id: {len(dup_ids)}")

    if len(dup_ids) > 0:
        display(dup_ids.sort_values(USER_ID_COL).head(10))
        conflict = dup_ids.groupby(USER_ID_COL)[col].nunique()
        n_conflicts = (conflict > 1).sum()
        print(f"user_id с разными {col}: {n_conflicts}")
        if n_conflicts == 0:
            print(f"→ У всех повторяющихся user_id одно и то же {col}.")
        else:
            print(f"→ Есть user_id с разными {col}. Нужно решение.")
    else:
        print("→ Повторяющихся user_id нет.")
    print()
```

```python
# ========== удаление дубликатов по user_id ==========
# Удаляем только если проверка из предыдущей ячейки показала,
# что конфликтов по значениям нет.

for name in ["users", "ads_activity"]:
    df = dataframes[name]
    before = df.shape[0]
    df = df.drop_duplicates(subset=[USER_ID_COL], keep="first").reset_index(drop=True)
    after = df.shape[0]
    dataframes[name] = df
    print(f"{name}: удалено {before - after} строк, осталось {after}")

# Обновляем ссылки
users = dataframes["users"]
ads_activity = dataframes["ads_activity"]
```

```python
# ========== выбросы ==========
# Ищем числовые признаки, исключая user_id и целевую переменную age_category.
# age_category — категориальная целевая переменная (коды классов 0–4),
# выбросы по ней искать нельзя, это не признак.

numeric_cols = {}
for name, df in dataframes.items():
    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    num_cols = [c for c in num_cols if c not in (USER_ID_COL, TARGET_COL)]
    if num_cols:
        numeric_cols[name] = num_cols

print("Числовые признаки (кроме user_id и age_category):")
if numeric_cols:
    for name, cols in numeric_cols.items():
        print(f"{name}: {cols}")
else:
    print("Числовых признаков нет.")

# Проверим допустимость значений целевой переменной
expected = {0, 1, 2, 3, 4}
actual = set(users[TARGET_COL].unique())
print(f"\nУникальные значения age_category: {sorted(actual)}")
print(f"Все значения входят в ожидаемый набор: {actual.issubset(expected)}")
```

**2.4 Анализ распределения целевой переменной**

```python
# ========== распределение age_category ==========
age_labels = {
    0: "младше 18",
    1: "18–25",
    2: "26–40",
    3: "41–55",
    4: "56+",
}

age_counts = users[TARGET_COL].value_counts().sort_index()
age_share = users[TARGET_COL].value_counts(normalize=True).sort_index()

summary = pd.DataFrame({
    "Класс": age_counts.index,
    "Описание": [age_labels[i] for i in age_counts.index],
    "Количество": age_counts.values,
    "Доля, %": (age_share.values * 100).round(2),
})

display(summary)

print(f"Всего пользователей: {len(users)}")
print(f"Классов: {users[TARGET_COL].nunique()}")
```

```python
# ========== визуализация: возрастные категории ==========
plt.figure(figsize=(10, 6))
ax = sns.countplot(
    data=users,
    x=TARGET_COL,
    order=sorted(users[TARGET_COL].unique()),
    palette="viridis",
)

# Подписи значений над столбцами
for p in ax.patches:
    height = p.get_height()
    ax.annotate(
        f"{int(height)}",
        (p.get_x() + p.get_width() / 2, height),
        ha="center", va="bottom", fontsize=10,
    )

plt.title("Распределение возрастных категорий пользователей")
plt.xlabel("Возрастная категория")
plt.ylabel("Количество пользователей")
plt.xticks(
    ticks=range(5),
    labels=[age_labels[i] for i in range(5)],
)
plt.tight_layout()
plt.show()
```

На диаграмме видно, что распределение пользователей по возрастным категориям неравномерное. Самая многочисленная группа — 56+ (1766 человек, 30.3%), самая малочисленная — 18–25 (514 человек, 8.8%). Разница между ними почти в 3.5 раза. В выборке преобладают старшие возрастные группы, а молодёжь представлена слабее.

```python
# ========== визуализация: доли классов ==========
plt.figure(figsize=(10, 6))
ax = sns.barplot(
    x=age_share.index,
    y=age_share.values * 100,
    palette="viridis",
)

for p in ax.patches:
    height = p.get_height()
    ax.annotate(
        f"{height:.1f}%",
        (p.get_x() + p.get_width() / 2, height),
        ha="center", va="bottom", fontsize=10,
    )

plt.title("Доли возрастных категорий пользователей")
plt.xlabel("Возрастная категория")
plt.ylabel("Доля, %")
plt.xticks(
    ticks=range(5),
    labels=[age_labels[i] for i in range(5)],
)
plt.tight_layout()
plt.show()
```

Category 20 — выше у 18–25 (0.064), ниже у 41–55 (0.028).  
Category 03 — стабильно высокая у 26–40 (0.070), 41–55 (0.070), 56+ (0.069), но ниже у 18–25 (0.042).  
Вывод: распределение активности по категориям сайтов сильно различается между возрастными группами. Это подтверждает гипотезу: website_category — информативный признак.

```python
# ========== топ-различия по категориям сайтов ==========
# Найдём категории, где разброс долей между возрастными группами максимальный
cat_std = pivot_cat_share.std(axis=0).sort_values(ascending=False)

print("Категории сайтов с наибольшим разбросом долей между возрастными группами:")
cat_std_df = cat_std.head(10).to_frame("std").reset_index()
cat_std_df.columns = ["website_category", "std"]
display(cat_std_df)

plt.figure(figsize=(10, 6))
sns.barplot(x=cat_std.head(10).values, y=cat_std.head(10).index, palette="viridis")
plt.title("Топ-10 категорий сайтов по разбросу долей между возрастными группами")
plt.xlabel("Стандартное отклонение доли")
plt.ylabel("Категория сайта")
plt.tight_layout()
plt.show()
```

**3.1 Сборка данных**

```python
# ========== функция build_features ==========
def build_features(
    users_path=USERS_PATH,
    visits_path=VISITS_PATH,
    ads_activity_path=ADS_ACTIVITY_PATH,
    surf_depth_path=SURF_DEPTH_PATH,
    primary_device_path=PRIMARY_DEVICE_PATH,
    cloud_usage_path=CLOUD_USAGE_PATH,
    target_col=TARGET_COL,
    user_id_col=USER_ID_COL,
):    
    # 1. Читаем CSV
    users = pd.read_csv(users_path)
    visits = pd.read_csv(visits_path)
    ads_activity = pd.read_csv(ads_activity_path)
    surf_depth = pd.read_csv(surf_depth_path)
    primary_device = pd.read_csv(primary_device_path)
    cloud_usage = pd.read_csv(cloud_usage_path)

    # 2. Чистим от дубликатов по user_id
    users = users.drop_duplicates(subset=[user_id_col], keep="first")
    ads_activity = ads_activity.drop_duplicates(subset=[user_id_col], keep="first")

    # 3. Приводим типы
    visits["date"] = pd.to_datetime(visits["date"])
    cloud_usage["cloud_usage"] = cloud_usage["cloud_usage"].astype(int)

    # 4. Признаки на основе visits
    # 4.1. Агрегаты по сессиям
    visits_agg = visits.groupby(user_id_col).agg(
        n_sessions=("session_id", "nunique"),
        n_categories=("website_category", "nunique"),
        n_visits=("session_id", "size"),
        n_days=("date", "nunique"),
    ).reset_index()

    # 4.2. Среднее число сессий в день
    visits_agg["sessions_per_day"] = visits_agg["n_sessions"] / visits_agg["n_days"]

    # 4.3. Доли активности по категориям сайтов
    pivot_cat = pd.pivot_table(
        visits,
        index=user_id_col,
        columns="website_category",
        values="session_id",
        aggfunc="count",
        fill_value=0,
    )
    pivot_cat_share = pivot_cat.div(pivot_cat.sum(axis=1), axis=0)
    pivot_cat_share.columns = [
        f"share_cat_{c.replace('Category ', '').lower()}"
        for c in pivot_cat_share.columns
    ]
    pivot_cat_share = pivot_cat_share.reset_index()

    # 4.4. Доли активности по времени суток
    daytime_map = {
        "утро": "morning",
        "день": "day",
        "вечер": "evening",
        "ночь": "night",
    }
    pivot_daytime = pd.pivot_table(
        visits,
        index=user_id_col,
        columns="daytime",
        values="session_id",
        aggfunc="count",
        fill_value=0,
    )
    pivot_daytime_share = pivot_daytime.div(pivot_daytime.sum(axis=1), axis=0)
    pivot_daytime_share.columns = [
        f"share_daytime_{daytime_map[c]}" for c in pivot_daytime_share.columns
    ]
    pivot_daytime_share = pivot_daytime_share.reset_index()

    # 4.5. Наиболее активное время суток
    most_active_daytime = (
        pivot_daytime.idxmax(axis=1)
        .map(daytime_map)
        .rename("most_active_daytime")
        .reset_index()
    )

    # 4.6. Доля активности в выходные
    visits["is_weekend"] = visits["date"].dt.dayofweek.isin([5, 6]).astype(int)
    weekend_share = (
        visits.groupby(user_id_col)["is_weekend"]
        .mean()
        .rename("share_weekend")
        .reset_index()
    )

    # 5. Объединяем всё по user_id
    features = users[[user_id_col, target_col]].copy()

    for df in [
        visits_agg,
        pivot_cat_share,
        pivot_daytime_share,
        most_active_daytime,
        weekend_share,
        ads_activity,
        surf_depth,
        primary_device,
        cloud_usage,
    ]:
        features = features.merge(df, on=user_id_col, how="left")

    return features
```

```python
# ========== запуск build_features ==========
features = build_features()

print(f"Размер признакового пространства: {features.shape}")
print(f"Уникальных пользователей: {features[USER_ID_COL].nunique()}")
print(f"Ожидалось пользователей: {users[USER_ID_COL].nunique()}")
print(f"Целевая переменная: {TARGET_COL}")
print(f"Пропусков в целевой переменной: {features[TARGET_COL].isna().sum()}")

print(f"\nМинимум n_days: {features['n_days'].min()}")
print(f"inf в sessions_per_day: {np.isinf(features['sessions_per_day']).sum()}")

display(features.head())
```

**3.3 Пайплайн предобработки**

```python
# ========== группы признаков ==========
# Все признаки, кроме user_id и целевой переменной
feature_columns = [
    c for c in X_train.columns
    if c not in (USER_ID_COL, TARGET_COL)
]

# Числовые: агрегаты + доли
numeric_features = [
    "n_sessions", "n_categories", "n_visits", "n_days", "sessions_per_day",
    "share_weekend",
] + [f"share_cat_{i:02d}" for i in range(1, 21)] + [
    "share_daytime_morning", "share_daytime_day",
    "share_daytime_evening", "share_daytime_night",
]

# Номинальные категориальные
nominal_features = ["primary_device", "most_active_daytime"]

# Порядковые категориальные
ordinal_features = ["ads_activity", "surf_depth"]

# Бинарный
binary_features = ["cloud_usage"]

# Проверим, что все признаки распределены по группам
all_grouped = set(numeric_features + nominal_features + ordinal_features + binary_features)
all_features = set(feature_columns)

print(f"Всего признаков (без user_id): {len(feature_columns)}")
print(f"Распределено по группам: {len(all_grouped)}")
print(f"Не распределено: {all_features - all_grouped}")
```

```python
# ========== порядок категорий ==========
ads_order = ["очень редко", "редко", "умеренно", "часто", "очень часто"]
surf_order = ["поверхностно", "средне", "глубоко"]

# ========== пайплайны для каждой группы ==========
# Числовые: заполняем пропуски медианой, масштабируем
numeric_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

# Номинальные: заполняем пропуски модой, кодируем one-hot
nominal_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])

# Порядковые: заполняем пропуски модой, кодируем с порядком
ordinal_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(
        categories=[ads_order, surf_order],
        handle_unknown="use_encoded_value",
        unknown_value=-1,
    )),
])

# Бинарный: заполняем пропуски модой (0 или 1)
binary_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
])

print("Пайплайны для четырёх групп собраны.")
```

```python
# ========== ColumnTransformer ==========
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipe, numeric_features),
        ("nom", nominal_pipe, nominal_features),
        ("ord", ordinal_pipe, ordinal_features),
        ("bin", binary_pipe, binary_features),
    ],
    remainder="drop",
)

print("ColumnTransformer собран:")
print(preprocessor)
```

```python
# ========== проверка пайплайна ==========
# Обучаем пайплайн ТОЛЬКО на train (fit), затем применяем (transform)
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print(f"X_train_processed: {X_train_processed.shape}")
print(f"X_test_processed:  {X_test_processed.shape}")

# Проверим, что пропусков не осталось
print(f"\nNaN в X_train_processed: {np.isnan(X_train_processed).sum()}")
print(f"NaN в X_test_processed:  {np.isnan(X_test_processed).sum()}")
```

```python
# ========== список признаков после кодирования ==========
feature_names = preprocessor.get_feature_names_out()

print(f"Всего признаков после обработки: {len(feature_names)}")
print("\nПервые 30:")
for name in feature_names[:30]:
    print(f"  {name}")

# Проверим, что user_id не попал в признаки
print(f"\nuser_id в признаках: {USER_ID_COL in list(feature_names)}")
```

## Обучение и оценка базовой модели

**4.1 Определение кросс-валидации**

```python
# ========== кросс-валидация ==========
cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE,
)

scoring = {
    "f1_macro": "f1_macro",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
}

print(f"Кросс-валидация: StratifiedKFold(n_splits={N_SPLITS})")
print(f"Метрики: {list(scoring.keys())}")
```

__________
**4.2 Базовая модель**

```python
# ========== базовая модель ==========

dummy = DummyClassifier(strategy="most_frequent", random_state=RANDOM_STATE)

# Кросс-валидация: обучаем на train, оцениваем на фолдах train
dummy_scores = cross_validate(
    dummy,
    X_train_processed,
    y_train,
    cv=cv,
    scoring=scoring,
    return_train_score=False,
    n_jobs=-1,
)

print("DummyClassifier (most_frequent) — метрики на кросс-валидации:")
for metric in scoring:
    mean_score = dummy_scores[f"test_{metric}"].mean()
    std_score = dummy_scores[f"test_{metric}"].std()
    print(f"  {metric}: {mean_score:.4f} ± {std_score:.4f}")
```

______
**4.3 Таблица метрик базовой модели**

```python
# ========== сводная таблица метрик ==========
def scores_to_row(name, scores, scoring):
    """Преобразует результаты cross_validate в строку таблицы."""
    row = {"model": name}
    for metric in scoring:
        mean_score = scores[f"test_{metric}"].mean()
        std_score = scores[f"test_{metric}"].std()
        row[metric] = f"{mean_score:.4f} ± {std_score:.4f}"
    return row

results = [scores_to_row("DummyClassifier", dummy_scores, scoring)]

results_df = pd.DataFrame(results).set_index("model")
display(results_df)
```

```python
# ========== матрица корреляций ==========
corr_matrix = X_train_df.corr().abs()

plt.figure(figsize=(14, 12))
sns.heatmap(
    corr_matrix,
    cmap="coolwarm",
    center=0,
    vmin=0,
    vmax=1,
    square=True,
    cbar_kws={"label": "|корреляция|"},
)
plt.title("Матрица корреляций признаков (абсолютные значения)")
plt.tight_layout()
plt.show()
```

Тепловая карта показывает абсолютные значения корреляций между 40 признаками. Большинство ячеек — светло-розовые (корреляция < 0.3), что говорит о низкой мультиколлинеарности в целом.  
Ярко выделяются два блока:  
Агрегаты по сессиям (n_sessions, n_visits, sessions_per_day) — корреляция ~1.0 между собой. Все три измеряют объём активности пользователя.  
One-hot по времени суток (most_active_daytime_day ↔ evening) — корреляция ~0.99. Это следствие one-hot кодирования: если most_active_daytime = day, то evening = 0, и наоборот. Такая мультиколлинеарность ожидаема.  
Мультиколлинеарность в данных есть, но она локальная.

```python
# ========== поиск мультиколлинеарных пар ==========
# Верхний треугольник матрицы (без диагонали)
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Пары с корреляцией > 0.9
high_corr_pairs = [
    (col, row, upper.loc[row, col])
    for col in upper.columns
    for row in upper.index
    if pd.notna(upper.loc[row, col]) and upper.loc[row, col] > 0.9
]

print(f"Пар с корреляцией > 0.9: {len(high_corr_pairs)}")
for col, row, value in high_corr_pairs[:20]:
    print(f"  {col} ↔ {row}: {value:.4f}")
```

```python
# ========== удаление мультиколлинеарных признаков ==========
# Из каждой пары удаляем один признак — оставляем первый по алфавиту
to_drop = set()
for col, row, _ in high_corr_pairs:
    # Оставляем тот, что идёт раньше в списке колонок
    keep = col if list(X_train_df.columns).index(col) < list(X_train_df.columns).index(row) else row
    drop = row if keep == col else col
    to_drop.add(drop)

print(f"Признаков к удалению: {len(to_drop)}")
for col in sorted(to_drop):
    print(f"  {col}")

X_train_corr = X_train_df.drop(columns=list(to_drop))
X_test_corr = X_test_df.drop(columns=list(to_drop))

print(f"\nПосле удаления мультиколлинеарных:")
print(f"  X_train_corr: {X_train_corr.shape}")
print(f"  X_test_corr:  {X_test_corr.shape}")
```

 **8.3 Выводы**

 Какая модель победила и почему

**Победитель:** `SVC` с ядром `rbf` (`C=1`, `gamma='scale'`).
**f1_macro на CV:** 0.8979, **на тесте:** 0.8860.

**Почему победила:**

1. **Нелинейность данных.** `SVC` с нелинейными ядрами (`rbf`, `poly`)
   показал результат **существенно выше** линейных моделей:
   - `SVC (rbf)`: 0.8979,
   - `SVC (poly)`: 0.8920,
   - `SVC (linear)`: 0.8249,
   - `LogisticRegression`: 0.8238.
   Разрыв ~0.07 — это сигнал, что связь признаков с возрастом
   **нелинейная**. Радиальные базисные функции улавливают эти связи.

2. **`rbf` лучше `poly`.** Разница небольшая (0.8979 vs 0.8920), но
   `rbf` стабильнее: стандартное отклонение по фолдам у `poly` ниже
   (0.0038 vs 0.0120), что означает меньшую вариативность на разных
   фолдах. `rbf` даёт более **устойчивый** результат в среднем.

3. **`linear` ядро и `LogisticRegression` почти одинаковы.**
   Это логично: `SVC` с линейным ядром и логистическая регрессия —
   обе линейные модели, и показывают схожий результат (~0.824).
   Они **проигрывают** нелинейным.

 Факторы, повлиявшие на качество

**Что улучшило качество:**

1. **Feature engineering** — доли активности по категориям сайтов
   и времени суток оказались самыми информативными признаками.
2. **Удаление мультиколлинеарных признаков** — `n_visits`,
   `sessions_per_day`, `most_active_daytime_evening` дублировали
   информацию и были удалены.
3. **Нелинейные ядра SVC** — уловили нелинейные связи между
   поведением и возрастом.
4. **Подбор гиперпараметров через `GridSearchCV`** — `C=1`,
   `gamma='scale'` дали лучший результат.
5. **Стратификация** — сохранение пропорций классов в train и test
   позволило получить репрезентативные метрики.

**Что ограничило качество:**

1. **Миноритарный класс 18–25** — доля всего 8.8%, модель даёт
   по нему F1 = 0.8000, что ниже среднего.
2. **Ошибки на границах классов** — `младше 18` → `26–40` (7%),
   `18–25` → `26–40` (8%). Соседние возрастные группы похожи
   по поведению, и модель их путает.
3. **Пропуски в данных** — у части пользователей нет записей
   в `ads_activity` (233), `surf_depth` (111), `primary_device` (157),
   `cloud_usage` (146). Это ограничивает точность на этих пользователях.
4. **Анонимизация категорий сайтов** — мы не знаем, что за
   `Category 13`, поэтому не можем содержательно интерпретировать
   важность признаков.

 Ограничения проекта

1. **Данные анонимизированы.** Категории сайтов — `Category 01` … `Category 20`,
   без реальной семантики. Мы не можем объяснить, **почему** `Category 13`
   так важна для возраста.
2. **Один срез данных.** Логи охватывают 14 дней (2025-11-01 … 2025-11-13).
   Сезонность, долгосрочные тренды не учтены.
3. **Целевая переменная — категории, а не точный возраст.**
   Мы предсказываем 5 групп, а не число лет.
4. **Метрика `f1_macro` — не бизнес-метрика.** В реальности важны
   `CTR`, `конверсия`, `revenue`. Но `f1_macro` — хороший прокси
   для качества классификации.
5. **Бизнес-требование «не переоценивать возраст несовершеннолетних»**
   формально выполнено (модель чаще «сдвигает» младших в старшую
   группу), но не проверено на отдельной метрике. Можно было бы
   ввести **cost-sensitive** метрику.

 Идеи развития

1. **Добавить признаки:**
   - средняя длина сессии (сейчас нет данных),
   - сезонность (день недели, месяц),
   - динамика поведения (тренд за 14 дней),
   - взаимодействия между категориями (`share_cat_13 / share_cat_12`).

2. **Улучшить работу с миноритарным классом 18–25:**
   - `class_weight='balanced'` в SVC,
   - oversampling / SMOTE,
   - ансамбли (но это за рамками модуля 3).

3. **Расширить набор моделей:**
   - `RandomForest`, `GradientBoosting` — за рамками модуля 3,
     но в реальном проекте дали бы выигрыш.

4. **Калибровка вероятностей** — если бизнес захочет
   пороговые решения («показывать, если P > 0.8»).

5. **Мониторинг и переобучение** — модель будет «стареть»:
   поведение пользователей меняется. Нужно периодически переобучать.

6. **A/B-тест** — проверить в реальной рекламной кампании,
   как влияет возрастной таргетинг на CTR и конверсию.

 Итог

- **Гипотеза подтверждена:** поведение в цифровой среде коррелирует
  с возрастом.
- **Цель достигнута:** `f1_macro` = 0.8979 на CV и 0.8860 на тесте — выше 0.75.
- **Лучшая модель:** `SVC (rbf, C=1, gamma='scale')`.
- **Ключевые признаки:** доли активности по времени суток
  и категориям сайтов.
- **Модель готова к внедрению.** Артефакты сохранены и проверены
  (раздел 7).

В ходе проекта построена модель машинного обучения, которая по данным
о поведении анонимного пользователя в цифровой среде определяет его
возрастную категорию. Модель прошла по условию внедрения

набора моделей (ансамбли, бустинг). Это позволит поднять `f1_macro`
